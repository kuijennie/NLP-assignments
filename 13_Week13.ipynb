```python
# Demonstrate lexical ambiguity using the word 'bank'
sentences = [
    "The bank approved my loan application.",
    "The children sat on the bank of the river."
]

for sentence in sentences:
    lower_sentence = sentence.lower()

    if "loan" in lower_sentence or "money" in lower_sentence:
        meaning = "Financial institution"
    elif "river" in lower_sentence or "water" in lower_sentence:
        meaning = "Side of a river"
    else:
        meaning = "Meaning is ambiguous"

    print("Sentence:", sentence)
    print("Interpreted meaning of 'bank':", meaning)
    print("-" * 60)
```

## 4. Mini Retrieval-Augmented Generation (RAG) Demonstration

RAG retrieves relevant information from a knowledge base before an answer is produced.


```python
# Create a small knowledge base for the RAG demonstration
knowledge_base = [
    "Large Language Models are used for chatbots, content generation, coding assistance and virtual tutoring.",
    "Multimodal AI combines text with other data types such as images, audio and video.",
    "AI agents can plan tasks, use tools and complete multi-step workflows.",
    "Retrieval-Augmented Generation retrieves relevant information from external documents before generating an answer.",
    "NLP in Africa can support local language translation, agriculture, healthcare, education and government services.",
    "Responsible NLP requires fairness, privacy, transparency, accountability and human oversight.",
    "Low-resource African languages often have limited datasets, digital text and language-processing tools."
]

# Convert the knowledge base into TF-IDF vectors
vectorizer = TfidfVectorizer(stop_words="english")
knowledge_vectors = vectorizer.fit_transform(knowledge_base)

print("Knowledge base prepared successfully.")
print("Number of documents:", len(knowledge_base))
```

```python
# Define a function that retrieves the most relevant information
def retrieve_information(question):
    question_vector = vectorizer.transform([question])

    similarity_scores = cosine_similarity(
        question_vector,
        knowledge_vectors
    ).flatten()

    best_index = similarity_scores.argmax()

    return knowledge_base[best_index], similarity_scores[best_index]


# Ask a sample question
question = "What opportunities does NLP provide in Africa?"

retrieved_text, score = retrieve_information(question)

print("USER QUESTION:")
print(question)

print("\nRETRIEVED INFORMATION:")
print(retrieved_text)

print("\nSIMILARITY SCORE:")
print(round(float(score), 4))
```

```python
# Test the retriever with several questions
test_questions = [
    "What is multimodal AI?",
    "Why is privacy important in NLP?",
    "How can NLP support Africa?",
    "What do AI agents do?"
]

print("RAG RETRIEVAL TEST RESULTS")
print("=" * 70)

for question in test_questions:
    retrieved_text, score = retrieve_information(question)

    print("\nQuestion:", question)
    print("Retrieved:", retrieved_text)
    print("Score:", round(float(score), 4))
    print("-" * 70)
```

## 5. Simple AI Agent Routing Demonstration

An AI agent may decide which tool or specialised component should handle a user request.


```python
# Define a simple AI agent router
def route_nlp_request(user_request):
    request = user_request.lower()

    if any(word in request for word in ["research", "information", "document", "knowledge"]):
        return "RAG Retriever"

    elif any(word in request for word in ["bias", "privacy", "fairness", "ethics"]):
        return "Ethics Reviewer"

    elif any(word in request for word in ["translate", "swahili", "language", "multilingual"]):
        return "Multilingual Assistant"

    else:
        return "General NLP Assistant"


# Test the routing agent
agent_requests = [
    "Find information from my research documents.",
    "Check whether this system may contain bias.",
    "Translate this sentence into Swahili.",
    "Summarize this paragraph."
]

print("AI AGENT ROUTING RESULTS")
print("=" * 60)

for request in agent_requests:
    selected_agent = route_nlp_request(request)

    print("Request:", request)
    print("Selected Agent:", selected_agent)
    print("-" * 60)
```

## 7. Ethical Considerations

Responsible NLP development should consider fairness, bias, privacy, transparency, accountability, safety and human oversight.


```python
# Define a simple ethical checklist for an NLP project
def ethical_nlp_check(
    uses_personal_data,
    tested_for_bias,
    has_human_oversight,
    explains_limitations
):
    checks = {
        "Privacy protection considered": not uses_personal_data,
        "Bias testing completed": tested_for_bias,
        "Human oversight available": has_human_oversight,
        "System limitations explained": explains_limitations
    }

    print("ETHICAL NLP CHECKLIST")
    print("=" * 50)

    passed = 0

    for item, status in checks.items():
        result = "PASS" if status else "NEEDS ATTENTION"
        print(f"{item}: {result}")

        if status:
            passed += 1

    print("\nChecks passed:", passed, "out of", len(checks))


# Evaluate a sample NLP project
ethical_nlp_check(
    uses_personal_data=False,
    tested_for_bias=True,
    has_human_oversight=True,
    explains_limitations=True
)
```
